```python
df.groupby('menu_item_name')['quantity_sold'].sum().sort_values(ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt
```

```python
top_item = df.groupby('menu_item_name')['quantity_sold'].sum().sort_values(ascending=False).head(10)
top_item.plot(kind = 'bar')
plt.title('Top Selling Menu Items')
plt.xlabel('Menu Item')
plt.ylabel('Quantity Sold')

plt.show()
```
