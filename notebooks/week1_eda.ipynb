```python
import pandas as pd
```

```python
df = pd.read_csv("../data/restaurant_sales_data.csv")
```

```python
df.info()
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df.info()
```

```python
df['quantity_sold'].describe()
```

```python
df.groupby('menu_item_name')['quantity_sold'].sum().sort_values(ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt
```

```python
top_items = df.groupby('menu_item_name')['quantity_sold'].sum().sort_values(ascending=False).head(10)
top_items.plot(kind = 'bar')
plt.title('Top Selling Menu Items')
plt.xlabel('Menu Item')
plt.ylabel('Quantity Sold')

plt.show()
```

```python
top_items.head(1)
```

```python
df['day_name'] = df['date'].dt.day_name()
```

```python
day_sales = df.groupby('day_name')['quantity_sold'].mean()
```

```python
day_sales
```

```python
day_sales.plot(kind = 'bar')

plt.title('Average Sales by day')
plt.xlabel('Day')
plt.ylabel('Average Quantity sold')
plt.show()
```

```python
day_sales.sort_values(ascending = False).head(1)
```

```python
day_sales.sort_values()
```

```python
df.groupby('has_promotion')['quantity_sold'].mean()
```

```python
promotion_sales = df.groupby('has_promotion')['quantity_sold'].mean()

promotion_sales.plot(kind='bar')

plt.title('Average Sales With vs Without Promotion')
plt.xlabel('Promotion')
plt.ylabel('Average Quantity Sold')

plt.show()
```

```python
weather_sales = df.groupby('weather_condition')['quantity_sold'].mean()
weather_sales
```

```python
weather_sales.plot(kind = 'bar')

plt.title('Average sale by weather')
plt.xlabel('weather')
plt.ylabel('Average Quantity Sold')

plt.show()
```

```python
event_sales = df.groupby('special_event')['quantity_sold'].mean()
event_sales 
```

```python
event_sales.plot(kind='bar')

plt.title('Average Sales During Special Events')
plt.xlabel('Special Event')
plt.ylabel('Average Quantity Sold')

plt.show()
```

```python
df['month'] = df['date'].dt.month_name()
```

```python
monthly_sales = df.groupby('month')['quantity_sold'].sum().sort_values(ascending = False)
monthly_sales
```

```python
monthly_sales.plot(kind='bar')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Quantity Sold')

plt.show()
```

```python
daily_sales = df.groupby('date')['quantity_sold'].sum()
```

```python
moving_avg = daily_sales.rolling(window=7).mean()
moving_avg
```

```python
plt.figure(figsize=(12,5))

daily_sales.plot(label='Daily Sales')

moving_avg.plot(label='7-Day Moving Average')

plt.title('Sales Trend with Moving Average')

plt.xlabel('Date')
plt.ylabel('Quantity Sold')

plt.legend()

plt.show()
```
